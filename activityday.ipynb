```python
from mini8.notebook_tools import CPU


cpu = CPU("""
LDI R0, 1   ;iterator
LDI R1, 20  ;maximum of 20
LDI R2, 2   ;interval value
CMP R0, R1  
JL print    ;if the iterator is less than 20 then jump to print
JGE finish
loop:
    CMP R0, R1  
    JL print
    JGE finish


print:
    OUT R0
    ADD R0, R2
    CMP R0, R1
    JGE finish
    JMP loop


finish:
    HALT
""")
#cpu = CPU(source, title="Lecture 2 practice")
cpu.run()
```

```python
from mini8.notebook_tools import CPU


cpu = CPU("""

LDI R0, 2
LDI R1, 0
LDI R2, 3
LDI R3, 1

JMP compare

check:
    JGE divide
    JL end
divide:
    ADD R1, R3
    SUB R0, R2
    JMP compare
compare:
    CMP R0, R2
    JMP check
end:
    OUT R1
    OUT R0
    HALT

""")
#cpu = CPU(source, title="Lecture 2 practice")
cpu.run()
```

```python
from mini8.notebook_tools import CPU


cpu = CPU("""

LDI R0, 4
LDI R1, 1

LDI R2, 0
CMP R0, R1
JL first
JGE second

first:
    MOV R2, R0
    JMP end
second:
    MOV R2, R1
    JMP end
end:
    OUT R2
    HALT

""")
#cpu = CPU(source, title="Lecture 2 practice")
cpu.run()
```
